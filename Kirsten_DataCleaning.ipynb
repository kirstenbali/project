#### Preserve Source and Inspect Columns and Types
Uses a function to load data and preserve original source file. Check the number of rows and columns in each file and the columns names and data types of each column.

```python
def import_file(file_name):
    df = pd.read_csv(file_name)
    data = df.copy()
    display(data.head(1))
    print("Number of rows:", len(data))
    print("Number of columns:", len(data.columns))
    print("Columns and Data Types:", data.dtypes.astype(str).to_dict())
    return data
```

```python
current_data = import_file('LAPD_NIBRS_Offenses_Dataset.csv')
historical_data = import_file('Crime_Data_from_2020_to_2024.csv')
```

#### Convert type of Date and Time Columns

```python
current_data = current_data.rename(columns={"Date Rptd": "Date_Reported", "Date OCC": "Date_Occurred", "Time OCC": "Time_Occurred"})
historical_data = historical_data.rename(columns={"Date Rptd": "Date_Reported", "DATE OCC": "Date_Occurred", "TIME OCC": "Time_Occurred"})
def clean_date_column(df, col):
    df["Date_"+col] = pd.to_datetime(df["Date_"+col], format="%Y %b %d %I:%M:%S %p", errors="coerce")
    df["Time_"+col] = df["Date_"+col].dt.time
    df["Date_"+col] = df["Date_"+col].dt.date
    return df
```

```python
current_cleaned_dates = clean_date_column(current_data, 'Reported')
current_cleaned_dates = clean_date_column(current_cleaned_dates, 'Occurred')
display(current_cleaned_dates.head(1))

historical_cleaned_dates = clean_date_column(historical_data, 'Reported')
historical_cleaned_dates = clean_date_column(historical_cleaned_dates, 'Occurred')
display(historical_cleaned_dates.head(1))
```

#### Standardized Text
Strip whitespaces in Area_Name and Location columns. Ensure letters are capitalized in UniqueNIBRNo. Makes sure the case numbers are either marked "Yes" or "no".

```python
current_data = current_cleaned_dates.rename(columns={
	"AREA": "Area_No",
	"AREA NAME": "Area_Name",
	"Crm Cd": "Crime_Code",
	"Crm Cd Desc": "Crime_Description",
	"Mocodes": "MO_Codes",
	"Vict Age": "Victim_Age",
	"Vict Sex": "Victim_Sex",
	"Vict Descent": "Victim_Descent",
	"LOCATION": "Location",
	"Premis Cd": "Premise_Code",
	"Premis Desc": "Premise_Description"
})

historical_data = historical_cleaned_dates.rename(columns={
	"AREA": "Area_No",
	"AREA NAME": "Area_Name",
	"Crm Cd": "Crime_Code",
	"Crm Cd Desc": "Crime_Description",
	"Mocodes": "MO_Codes",
	"Vict Age": "Victim_Age",
	"Vict Sex": "Victim_Sex",
	"Vict Descent": "Victim_Descent",
	"LOCATION": "Location",
	"Premis Cd": "Premise_Code",
	"Premis Desc": "Premise_Description"
})

current_data.columns = current_data.columns.str.strip()
historical_data.columns = historical_data.columns.str.strip()

current_data["Area_Name"] = current_data["Area_Name"].str.strip()
historical_data["Area_Name"] = historical_data["Area_Name"].str.strip()
current_data["Location"] = current_data["Location"].str.replace(r'\s+', ' ', regex=True)
historical_data["Location"] = historical_data["Location"].str.replace(r'\s+', ' ', regex=True).str.title()
historical_data["Cross Street"] = historical_data["Cross Street"].str.strip()
```

#### Checking Suspicous Values

```python
def find_report_before_occurrence(df, dataset_name, id_column):
    """Return rows whose report timestamp is earlier than occurrence timestamp."""
    report_timestamp = pd.to_datetime(
        df["Date_Reported"].astype("string") + " "
        + df["Time_Reported"].astype("string"),
        errors="coerce",
    )
    occurrence_timestamp = pd.to_datetime(
        df["Date_Occurred"].astype("string") + " "
        + df["Time_Occurred"].astype("string"),
        errors="coerce",
    )

    comparable = report_timestamp.notna() & occurrence_timestamp.notna()
    report_precedes_occurrence = (
        comparable & report_timestamp.lt(occurrence_timestamp)
    )
    df["report_precedes_occurrence"] = report_precedes_occurrence

    result_columns = [
        column for column in [
            id_column, "Date_Reported", "Time_Reported",
            "Date_Occurred", "Time_Occurred",
        ]
        if column in df.columns
    ]
    flagged_rows = df.loc[report_precedes_occurrence, result_columns].copy()
    flagged_rows.insert(0, "dataset", dataset_name)
    flagged_rows["report_timestamp"] = report_timestamp.loc[report_precedes_occurrence]
    flagged_rows["occurrence_timestamp"] = occurrence_timestamp.loc[report_precedes_occurrence]

    print(f"{dataset_name}: comparable rows = {int(comparable.sum()):,}")
    print(f"{dataset_name}: report precedes occurrence = {len(flagged_rows):,}")
    return flagged_rows


report_before_occurrence = pd.concat(
    [
        find_report_before_occurrence(current_data, "NIBRS", "UniqueNIBRNo"),
        find_report_before_occurrence(historical_data, "Legacy", "DR_NO"),
    ],
    ignore_index=True,
)

# Review these records; do not automatically change their dates or remove them.
display(report_before_occurrence.head(25))
```

```python
numeric_columns = [
    "LAT",
    "LON",
    "TotalVictimCount",
    "Area_No",
    "RPT Dist No"
]
numeric_columns2 = [
    "LAT",
    "LON",
    "Area_No",
    "Rpt Dist No"
]

def convert_to_numeric(df, columns):
    for column in columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )
    return df

current_data = convert_to_numeric(current_data, numeric_columns)
historical_data = convert_to_numeric(historical_data, numeric_columns2)
print("Current data numeric columns: " + str(current_data[numeric_columns].dtypes.astype(str).to_dict()))
print("Historical data numeric columns: " + str(historical_data[numeric_columns2].dtypes.astype(str).to_dict()))
```

```python
# validate coordinates
def validate_coordinates(df):
    df["flag_missing_coordinates"] = (df["LAT"].isna() | df["LON"].isna())

    df["flag_zero_coordinates"] = (df["LAT"].eq(0) | df["LON"].eq(0))

    df["flag_coordinates_outside_la"] = (df["LAT"].notna() & df["LON"].notna() 
                                         & ~df["LAT"].eq(0) & ~df["LON"].eq(0)
        & (~df["LAT"].between(33.5, 34.9) | ~df["LON"].between(-119.0, -117.5)))

    df["valid_coordinates"] = (~df["flag_missing_coordinates"] & ~df["flag_zero_coordinates"]
                                & ~df["flag_coordinates_outside_la"])
    print("Rows with missing coordinates:")
    display(df[df["flag_missing_coordinates"]])
    print("Rows with zero coordinates:")
    display(df[df["flag_zero_coordinates"]])
    print("Rows with coordinates outside LA:")
    display(df[df["flag_coordinates_outside_la"]])

validate_coordinates(current_data)
validate_coordinates(historical_data)
```

```python
def validate_victim_count(df):
    df["flag_negative_victim_count"] = (df["TotalVictimCount"] < 0)

    df["flag_noninteger_victim_count"] = (
        df["TotalVictimCount"].notna()
        & (df["TotalVictimCount"] != df["TotalVictimCount"].round())
    )

    print("Rows with negative victim count:" + str(df["flag_negative_victim_count"].sum()))
    print("Rows with non-integer victim count:" + str(df["flag_noninteger_victim_count"].sum()))
validate_victim_count(current_data)
```

```python
def validate_area(df):
    df["flag_missing_area"] = df["Area_No"].isna()

    df["flag_invalid_area"] = (df["Area_No"].notna() & ~df["Area_No"].between(1, 21))

    df["flag_noninteger_area"] = (df["Area_No"].notna() & (df["Area_No"] != df["Area_No"].round()))

    print("Rows with missing area:" + str(df["flag_missing_area"].sum()))
    print("Rows with invalid area:" + str(df["flag_invalid_area"].sum()))
    print("Rows with non-integer area:" + str(df["flag_noninteger_area"].sum()))
validate_area(current_data)
validate_area(historical_data)
```

```python
def validate_district(df, name):
    df["flag_missing_district"] = (df[name].isna())

    df["flag_invalid_district"] = (df[name].notna() 
                                   & ~df[name].between(0, 2199))

    df["flag_noninteger_district"] = (df[name].notna() 
                                      & (df[name] != df[name].round()))

    df["district_area"] = (df[name] // 100)

    df["flag_area_district_mismatch"] = (df["Area_No"].notna() & df[name].notna()
                                          & (df["Area_No"] != df["district_area"]))

    print("Rows with missing district:" + str(df["flag_missing_district"].sum()))
    print("Rows with invalid district:" + str(df["flag_invalid_district"].sum()))

validate_district(current_data, "RPT Dist No")
validate_district(historical_data, "Rpt Dist No")
```

```python
validation_only_columns = {
    "report_precedes_occurrence",
    "valid_coordinates",
    "district_area",
}


def shareable_preview(df):
    columns = [
        column
        for column in df.columns
        if not column.startswith("flag_") and column not in validation_only_columns
    ]
    return df.loc[:, columns].head(50).copy()


current_shareable_preview = shareable_preview(current_data)
historical_shareable_preview = shareable_preview(historical_data)

print(f"Current cleaned data: showing {len(current_shareable_preview)} rows and {len(current_shareable_preview.columns)} shareable columns")
display(current_shareable_preview)

print(f"Historical cleaned data: showing {len(historical_shareable_preview)} rows and {len(historical_shareable_preview.columns)} shareable columns")
display(historical_shareable_preview)
```
